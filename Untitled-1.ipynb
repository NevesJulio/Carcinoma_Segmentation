```python
from pathlib import Path
import xml.etree.ElementTree as ET

import openslide
from PIL import ImageDraw
import matplotlib.pyplot as plt


# ============================================================
# CONFIGURAÇÕES
# ============================================================

PASTA_SVS = Path(
    r"/run/media/julio-neves/HD_JÚLIO/Laminas_ROI/HER2-ROI/PKG - HER2 tumor ROIs_v3/Yale_HER2_cohort/SVS"
)

PASTA_ROI = Path(
    r"/run/media/julio-neves/HD_JÚLIO/Laminas_ROI/HER2-ROI/PKG - HER2 tumor ROIs_v3/Yale_HER2_cohort/Annotations"
)

THUMB_MAX_SIZE = (2000, 2000)

MARGEM = 50
MIN_ZOOM = 200


# ============================================================
# FUNÇÃO PARA PROCESSAR UMA LÂMINA
# ============================================================

def processar_lamina(svs_path, xml_path):

    print("\n" + "=" * 70)
    print(f"Lâmina: {svs_path.name}")
    print(f"XML:    {xml_path.name}")

    slide = openslide.OpenSlide(str(svs_path))

    try:

        # ====================================================
        # DIMENSÕES
        # ====================================================

        largura_original, altura_original = slide.dimensions

        print(
            f"Dimensão original: "
            f"{largura_original} x {altura_original}"
        )


        # ====================================================
        # THUMBNAIL
        # ====================================================

        thumb_original = slide.get_thumbnail(
            THUMB_MAX_SIZE
        ).convert("RGB")

        thumb_anotada = thumb_original.copy()

        largura_thumb, altura_thumb = thumb_original.size

        print(
            f"Thumbnail: "
            f"{largura_thumb} x {altura_thumb}"
        )


        # ====================================================
        # ESCALA
        # ====================================================

        scale_x = largura_thumb / largura_original
        scale_y = altura_thumb / altura_original


        # ====================================================
        # ABRIR XML
        # ====================================================

        tree = ET.parse(xml_path)
        root = tree.getroot()


        # ====================================================
        # DESENHAR ROIs
        # ====================================================

        draw = ImageDraw.Draw(thumb_anotada)

        numero_rois = 0
        todos_pontos = []


        for region in root.iter("Region"):

            pontos = []

            # Primeiro procura Coordinate
            coordenadas = list(
                region.iter("Coordinate")
            )

            # Se não existir, tenta Vertex
            if not coordenadas:
                coordenadas = list(
                    region.iter("Vertex")
                )


            for coordinate in coordenadas:

                if "X" not in coordinate.attrib:
                    continue

                if "Y" not in coordinate.attrib:
                    continue

                x = float(
                    coordinate.attrib["X"]
                )

                y = float(
                    coordinate.attrib["Y"]
                )

                x_thumb = x * scale_x
                y_thumb = y * scale_y

                ponto = (
                    x_thumb,
                    y_thumb
                )

                pontos.append(ponto)
                todos_pontos.append(ponto)


            if len(pontos) >= 2:

                pontos_fechados = (
                    pontos + [pontos[0]]
                )

                draw.line(
                    pontos_fechados,
                    fill="red",
                    width=5
                )

                numero_rois += 1


        print(
            f"ROIs encontradas: {numero_rois}"
        )


        # ====================================================
        # SE NÃO ENCONTROU ROI
        # ====================================================

        if not todos_pontos:

            print(
                "Nenhuma coordenada encontrada. Pulando."
            )

            return


        # ====================================================
        # BOUNDING BOX
        # ====================================================

        xs = [
            p[0]
            for p in todos_pontos
        ]

        ys = [
            p[1]
            for p in todos_pontos
        ]

        x_min = min(xs)
        x_max = max(xs)

        y_min = min(ys)
        y_max = max(ys)


        # ====================================================
        # MARGEM
        # ====================================================

        x_min = max(
            0,
            x_min - MARGEM
        )

        x_max = min(
            largura_thumb,
            x_max + MARGEM
        )

        y_min = max(
            0,
            y_min - MARGEM
        )

        y_max = min(
            altura_thumb,
            y_max + MARGEM
        )


        # ====================================================
        # ZOOM MÍNIMO
        # ====================================================

        largura_roi = (
            x_max - x_min
        )

        altura_roi = (
            y_max - y_min
        )


        if largura_roi < MIN_ZOOM:

            centro_x = (
                x_min + x_max
            ) / 2

            x_min = (
                centro_x
                - MIN_ZOOM / 2
            )

            x_max = (
                centro_x
                + MIN_ZOOM / 2
            )


        if altura_roi < MIN_ZOOM:

            centro_y = (
                y_min + y_max
            ) / 2

            y_min = (
                centro_y
                - MIN_ZOOM / 2
            )

            y_max = (
                centro_y
                + MIN_ZOOM / 2
            )


        # Limites finais
        x_min = max(
            0,
            x_min
        )

        x_max = min(
            largura_thumb,
            x_max
        )

        y_min = max(
            0,
            y_min
        )

        y_max = min(
            altura_thumb,
            y_max
        )


        # ====================================================
        # PLOT
        # ====================================================

        fig, axes = plt.subplots(
            1,
            2,
            figsize=(16, 8)
        )


        # Original
        axes[0].imshow(
            thumb_original
        )

        axes[0].set_title(
            f"{svs_path.stem}\nOriginal"
        )

        axes[0].set_xlim(
            x_min,
            x_max
        )

        axes[0].set_ylim(
            y_max,
            y_min
        )

        axes[0].axis("off")


        # Anotada
        axes[1].imshow(
            thumb_anotada
        )

        axes[1].set_title(
            f"{svs_path.stem}\n"
            f"Com anotação - {numero_rois} ROI(s)"
        )

        axes[1].set_xlim(
            x_min,
            x_max
        )

        axes[1].set_ylim(
            y_max,
            y_min
        )

        axes[1].axis("off")


        plt.tight_layout()
        plt.show()


    finally:

        slide.close()


# ============================================================
# PERCORRER TODAS AS LÂMINAS
# ============================================================

arquivos_svs = sorted(
    PASTA_SVS.glob("*.svs")
)

print(
    f"{len(arquivos_svs)} lâminas encontradas."
)


quantidade_processada = 0
quantidade_sem_xml = 0


for svs_path in arquivos_svs:

    xml_path = (
        PASTA_ROI
        / f"{svs_path.stem}.xml"
    )


    # Se não existir XML com mesmo nome
    if not xml_path.exists():

        print(
            f"\nSem XML: {svs_path.name}"
        )

        quantidade_sem_xml += 1

        continue


    try:

        processar_lamina(
            svs_path,
            xml_path
        )

        quantidade_processada += 1


    except Exception as erro:

        print(
            f"\nErro em {svs_path.name}:"
        )

        print(erro)


# ============================================================
# RESUMO
# ============================================================

print("\n" + "=" * 70)

print(
    f"Processadas: {quantidade_processada}"
)

print(
    f"Sem XML correspondente: {quantidade_sem_xml}"
)
```
